## **General CMake**

Eigen is a **header**, so it only needs **include**\
Ceres, g2o, Sophus are **library**, so **include** and **lib**

```python
include_directories(/usr/include/eigen3 /usr/local/include)
link_directories(/usr/local/lib)

target_link_libraries(my_target
  ceres
  g2o_core g2o_stuff
  sophus
  glog gflags cholmod lapack blas
)
```
